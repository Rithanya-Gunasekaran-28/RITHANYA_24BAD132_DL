```python

print("RITHANYA.G 24BAD132")

def chatbot_response(user_input, history=None):
    if not user_input.strip():
        return "Please enter a question."

    messages = [
        {
            "role": "system",
            "content": (
                "You are a helpful educational AI assistant. "
                "Answer questions accurately in clear, simple English. "
                "Explain concepts with definitions, key points, and examples "
                "when useful. Avoid vague, repetitive, or one-word answers. "
                "If you are unsure, say so honestly."
            )
        }
    ]

    if history:
        for item in history[-4:]:
            if isinstance(item, dict):
                if item.get("role") in ("user", "assistant"):
                    messages.append({
                        "role": item["role"],
                        "content": item["content"]
                    })
            elif isinstance(item, (list, tuple)) and len(item) == 2:
                if item[0] and item[1]:
                    messages.append({"role": "user", "content": item[0]})
                    messages.append({"role": "assistant", "content": item[1]})

    messages.append({"role": "user", "content": user_input})

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=True,
            temperature=0.5,
            top_p=0.9,
            repetition_penalty=1.1,
            pad_token_id=tokenizer.eos_token_id
        )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    return tokenizer.decode(new_tokens, skip_special_tokens=True).strip()

print(chatbot_response("What is Artificial Intelligence?"))
print(chatbot_response("Explain machine learning with an example."))
```

```python

print("RITHANYA.G 24BAD132")

!pip -q install -U transformers accelerate

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)

def chatbot_response(user_input, history=None):
    if not user_input or not user_input.strip():
        return "Please enter a question."

    messages = [
        {
            "role": "system",
            "content": "You are a helpful educational assistant. Explain concepts accurately in simple English with definitions and examples."
        }
    ]

    for item in (history or [])[-4:]:
        if isinstance(item, dict) and item.get("role") in ("user", "assistant"):
            messages.append({
                "role": item["role"],
                "content": item["content"]
            })
        elif isinstance(item, (list, tuple)) and len(item) == 2:
            if item[0] and item[1]:
                messages.append({"role": "user", "content": item[0]})
                messages.append({"role": "assistant", "content": item[1]})

    messages.append({"role": "user", "content": user_input})

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False,
            repetition_penalty=1.1
        )

    answer_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    return tokenizer.decode(answer_tokens, skip_special_tokens=True).strip()

print("Model and tokenizer loaded successfully.")
print(chatbot_response("What is Artificial Intelligence?"))
```

```python

print("RITHANYA.G 24BAD132")

import gradio as gr

def chat(message, history):
    return chatbot_response(message, history or [])

demo = gr.ChatInterface(
    fn=chat,
    title="LLM Chatbot - RITHANYA.G 24BAD132",
    description="Ask questions and receive detailed AI-generated answers."
)

demo.launch(share=False)
```
